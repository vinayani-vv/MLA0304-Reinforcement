```python
states = ["A", "B", "Goal"]
cost = {"A": 5, "B": 2, "Goal": 0}

V = {"A": 0, "B": 0, "Goal": 0}

for _ in range(5):
    V["B"] = cost["B"] + V["Goal"]
    V["A"] = cost["A"] + min(V["B"], V["Goal"])

print("Optimal Cost from A:", V["A"])
print("Optimal Cost from B:", V["B"])
```
